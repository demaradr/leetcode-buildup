## Many ways to solve

| Approach | Idea | Time |
|----------|------|------|
| **Brute force** | Try every buy day `i` and every sell day `j > i`; keep the best profit | O(n²) |
| **Sliding window** (what you'll build) | `l` = buy day, `r` = sell day; slide right, move buy when a cheaper price appears | O(n) |

Both find the answer. **Sliding window is the classic pattern** — one pass, constant extra space.

## Why sliding window?

You may buy on **one** day and sell on a **later** day. Put `l` on the buy day and `r` on the sell day:

```python
l, r = 0, 1
```

On each step (`while r < len(prices)`):
1. If `prices[l] < prices[r]` → selling is profitable; update max profit with `prices[r] - prices[l]`
2. Else → today's price is cheaper (or equal); move `l = r` so you buy at this better price
3. Always `r += 1`

`l` never goes past `r`, and you only sell after you buy.
