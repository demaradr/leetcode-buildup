## Many ways to solve

| Approach | Idea | Time |
|----------|------|------|
| **Brute force** | Try every buy day `i` and every sell day `j > i` | O(n²) |
| **Sliding window** (below) | `l` = buy, `r` = sell; move buy when a cheaper price appears | O(n) |

**Sliding window is the classic pattern** — O(n) time, O(1) extra space.

## Why move `l`?

If `prices[r] <= prices[l]`, selling today cannot beat buying today instead. So set `l = r` and keep looking for a later sell day.

### Exercise 8 — One full window step with prints


```python
prices = [7, 1, 5, 3, 6, 4]
l, r = 0, 1
max_profit = 0

while r < len(prices):
    if prices[l] < prices[r]:
        profit = prices[r] - prices[l]
        max_profit = max(max_profit, profit)
        print("buy", prices[l], "sell", prices[r], "profit", profit)
    else:
        l = r
        print("moved buy to day", l, "price", prices[l])
    r += 1

print(max_profit)
```

### Exercise 9 — Put it in a function


```python
def max_profit(prices):
    l, r = 0, 1
    best = 0

    while r < len(prices):
        if prices[l] < prices[r]:
            best = max(best, prices[r] - prices[l])
        else:
            l = r
        r += 1

    return best


print(max_profit([7, 1, 5, 3, 6, 4]))
print(max_profit([7, 6, 4, 3, 1]))
print(max_profit([2, 4, 1]))
```

## Part 2 — Full solution

O(n) sliding-window solution: `l` tracks the best buy day so far, `r` walks every later sell day.


```python
class Solution:
    def maxProfit(self, prices: list[int]) -> int:
        l, r = 0, 1
        max_profit = 0

        while r < len(prices):
            if prices[l] < prices[r]:
                profit = prices[r] - prices[l]
                max_profit = max(max_profit, profit)
            else:
                l = r
            r += 1

        return max_profit
```

```python
sol = Solution()

assert sol.maxProfit([7, 1, 5, 3, 6, 4]) == 5
assert sol.maxProfit([7, 6, 4, 3, 1]) == 0
assert sol.maxProfit([2, 4, 1]) == 2
assert sol.maxProfit([1]) == 0
assert sol.maxProfit([1, 2]) == 1

print("All tests passed.")
```
