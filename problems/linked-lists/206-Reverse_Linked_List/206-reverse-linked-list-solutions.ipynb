## Many ways to solve

| Approach | Idea | Time / Space |
|----------|------|--------------|
| **Iterative** (below) | Flip each `next` to point at `prev` | O(n) time, O(1) space |
| **Recursive** | Reverse the rest, then point the next node back | O(n) time, O(n) stack |

**The iterative approach is the best space complexity** — one pass, constant extra memory.

## Why save `nxt`?

Before you set `curr.next = prev`, `curr.next` is the only reference to the rest of the list. Store it in `nxt`, flip the arrow, then slide `prev` and `curr` forward.


---
## Setup — `ListNode` helpers


```python
from typing import Optional, List


class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

    def __repr__(self):
        return f"ListNode({self.val})"


def build_list(values: List[int]) -> Optional[ListNode]:
    dummy = ListNode()
    curr = dummy
    for v in values:
        curr.next = ListNode(v)
        curr = curr.next
    return dummy.next


def to_list(head: Optional[ListNode]) -> List[int]:
    out = []
    while head:
        out.append(head.val)
        head = head.next
    return out


def print_list(head: Optional[ListNode]) -> None:
    print(" → ".join(str(v) for v in to_list(head)) if head else "None")
```

## Part 1 — Skill exercise solutions


### Exercise 1 — Create a single `ListNode`


```python
node = ListNode(1)
print(node.val)
print(node.next)
```

### Exercise 2 — Link two nodes by hand


```python
a = ListNode(1)
b = ListNode(2)
a.next = b

print(a.val)
print(a.next.val)
print(a.next.next)
```

### Exercise 3 — Build a list with the helper


```python
head = build_list([1, 2, 3])
print_list(head)
```

### Exercise 4 — Walk a list with `while curr`


```python
head = build_list([1, 2, 3])
curr = head

while curr:
    print(curr.val)
    curr = curr.next
```

### Exercise 5 — Save `next` before overwriting


```python
head = build_list([1, 2, 3])
curr = head

nxt = curr.next
curr.next = None
print(nxt.val)
print(to_list(curr))
```

### Exercise 6 — Flip one arrow: `curr.next = prev`


```python
head = build_list([1, 2, 3])
prev = None
curr = head

nxt = curr.next
curr.next = prev
print(to_list(curr))
print(nxt.val)
```

### Exercise 7 — Advance `prev` and `curr`


```python
head = build_list([1, 2, 3])
prev = None
curr = head

nxt = curr.next
curr.next = prev
prev = curr
curr = nxt

print(prev.val)
print(curr.val)
```

### Exercise 8 — One full reverse loop


```python
head = build_list([1, 2, 3])
prev, curr = None, head

while curr:
    nxt = curr.next
    curr.next = prev
    prev = curr
    curr = nxt

print_list(prev)
```

### Exercise 9 — Edge cases


```python
for values in ([], [1]):
    head = build_list(values)
    prev, curr = None, head

    while curr:
        nxt = curr.next
        curr.next = prev
        prev = curr
        curr = nxt

    print(to_list(prev))
```

### Exercise 10 — Wrap it in a function


```python
def reverse_list(head):
    prev, curr = None, head

    while curr:
        nxt = curr.next
        curr.next = prev
        prev = curr
        curr = nxt

    return prev


print(to_list(reverse_list(build_list([1, 2, 3, 4, 5]))))
print(to_list(reverse_list(build_list([1, 2]))))
print(to_list(reverse_list(build_list([]))))
```

## Part 2 — Full solution

O(n) iterative reverse — save `nxt`, flip `curr.next` to `prev`, slide both pointers, return `prev` as the new head.


```python
from typing import Optional


class Solution:
    def reverseList(self, head: Optional[ListNode]) -> Optional[ListNode]:
        prev, curr = None, head

        while curr:
            nxt = curr.next
            curr.next = prev
            prev = curr
            curr = nxt
        return prev
```

```python
sol = Solution()

assert to_list(sol.reverseList(build_list([1, 2, 3, 4, 5]))) == [5, 4, 3, 2, 1]
assert to_list(sol.reverseList(build_list([1, 2]))) == [2, 1]
assert to_list(sol.reverseList(build_list([]))) == []
assert to_list(sol.reverseList(build_list([1]))) == [1]

print("All tests passed.")
```
