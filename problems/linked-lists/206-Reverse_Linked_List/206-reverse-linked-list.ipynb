## Many ways to solve

| Approach | Idea | Time / Space |
|----------|------|--------------|
| **Iterative** (what you'll build) | Walk the list; flip each `next` to point at the previous node | O(n) time, O(1) space |
| **Recursive** | Reverse the rest of the list, then point the next node back at you | O(n) time, O(n) stack |

Both reverse the list correctly. **The iterative approach is the one to master** — constant extra space, and the same `prev` / `curr` pattern shows up in reverse-in-pairs, reverse-k-group, and other list problems.

## Why `prev` and `curr`?

A singly-linked list only points **forward**. To reverse it, each node must point at the node that *used* to come before it.

You need three ideas at each step:

1. **`curr`** — the node you are flipping right now
2. **`prev`** — the node that should become `curr.next` (the reverse direction)
3. **`nxt`** — save `curr.next` *before* you overwrite it, or you lose the rest of the list

```python
# head → 1 → 2 → 3 → None

prev, curr = None, head

# step on 1: nxt = 2; 1.next = None; prev = 1; curr = 2
# step on 2: nxt = 3; 2.next = 1; prev = 2; curr = 3
# step on 3: nxt = None; 3.next = 2; prev = 3; curr = None
# return prev  →  3 → 2 → 1 → None
```

Pattern you'll use:

```python
prev, curr = None, head
while curr:
    nxt = curr.next      # save the rest of the list
    curr.next = prev    # flip the arrow
    prev = curr         # slide prev forward
    curr = nxt          # slide curr forward
return prev             # new head
```

## Setup — `ListNode` helpers

Run this cell once. Every exercise below uses these helpers to build and print lists.


````python
from typing import Optional, List


class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

    def __repr__(self):
        return f"ListNode({self.val})"


def build_list(values: List[int]) -> Optional[ListNode]:
    """Turn [1, 2, 3] into 1 → 2 → 3 → None."""
    dummy = ListNode()
    curr = dummy
    for v in values:
        curr.next = ListNode(v)
        curr = curr.next
    return dummy.next


def to_list(head: Optional[ListNode]) -> List[int]:
    """Turn a linked list back into a Python list (for asserts / printing)."""
    out = []
    while head:
        out.append(head.val)
        head = head.next
    return out


def print_list(head: Optional[ListNode]) -> None:
    print(" → ".join(str(v) for v in to_list(head)) if head else "None")
````
