## Many ways to solve

| Approach | Idea | Time |
|----------|------|------|
| **Count + sort** | Count frequencies, sort unique numbers by count, take first `k` | O(n log n) |
| **Count + heap** (what you'll build) | Count frequencies, keep a min-heap of size `k` | O(n log k) |
| **Count + buckets** | Count frequencies, put numbers into buckets by count, walk from highest count down | O(n) |

Sorting works and is fine to understand first. The follow-up asks for **better than O(n log n)** — a size-`k` heap gives you O(n log k).

Buckets can reach O(n), but that trick is fairly specific to "frequency is bounded by `n`." **A heap of size `k` shows up everywhere** — kth largest, merge k lists, closest points, scheduling. Learn that pattern here.

## Why a min-heap of size `k`?

After you count, you have pairs like `(frequency, number)`. You only need the `k` highest frequencies.

Python's `heapq` is a **min-heap**: the smallest item sits at the front. So:

1. Push `(count, number)` for each unique number.
2. If the heap grows past `k`, **pop** — that removes the current *least* frequent item.
3. Whatever is left are the `k` most frequent numbers.

```python
# count = {1: 3, 2: 2, 3: 1}, k = 2

heap = []
# push (3, 1) → [(3, 1)]
# push (2, 2) → [(2, 2), (3, 1)]
# push (1, 3) → [(1, 3), (3, 1), (2, 2)]  then pop → [(2, 2), (3, 1)]
# heap holds the top 2: numbers 1 and 2
```

Pattern you'll use:

```python
count[n] = 1 + count.get(n, 0)   # build frequencies
heapq.heappush(heap, (c, n))    # push (frequency, number)
if len(heap) > k:
    heapq.heappop(heap)         # drop the least frequent
```
