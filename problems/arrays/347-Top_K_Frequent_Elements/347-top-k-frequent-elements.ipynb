## Many ways to solve

| Approach | Idea | Time |
|----------|------|------|
| **Count + sort** | Count frequencies, sort unique numbers by count, take first `k` | O(n log n) |
| **Count + heap** | Count frequencies, keep a min-heap of size `k` | O(n log k) |
| **Count + buckets** (what you'll build) | Count frequencies, put numbers into buckets by count, walk from highest count down | O(n) |

Sorting works and is fine to understand first. The follow-up asks for **better than O(n log n)**. **The bucket approach is the best time complexity** — O(n) time.

## Why buckets?

After you count, you know each number's frequency. Frequency is an integer between `1` and `n` (you can't appear more times than the array length).

So you can build an array of buckets where **the index is the frequency**:

```python
# nums = [1,1,1,2,2,3]  →  count = {1: 3, 2: 2, 3: 1}

freq = [
  [],      # index 0 — unused (nothing appears 0 times)
  [3],     # index 1 — numbers that appear once
  [2],     # index 2 — numbers that appear twice
  [1],     # index 3 — numbers that appear three times
  [],      # …
]
```

Then walk **from the end** (highest frequency) toward the start, appending numbers until you have `k` of them.

Pattern you'll use:

```python
count[n] = 1 + count.get(n, 0)   # build frequencies
freq[c].append(n)               # put number n into bucket c
```
