## Many ways to solve

| Approach | Idea | Time |
|----------|------|------|
| **Count + sort** | Sort unique numbers by frequency | O(n log n) |
| **Count + heap** | Min-heap of size `k` | O(n log k) |
| **Count + buckets** (below) | Index = frequency; walk high → low | O(n) |

**The bucket approach is the best time complexity** — O(n) time, O(n) space. It satisfies the follow-up (better than O(n log n)).

## Why buckets?

A number's frequency is at most `n`. Put each number into `freq[count]` so the **index is how often it appears**. Walking from the end collects the most frequent numbers first without sorting.

### Exercise 1 — Count frequencies with `.get()`


```python
nums = [1, 1, 1, 2, 2, 3]
count = {}

for n in nums:
    count[n] = 1 + count.get(n, 0)

print(count)
```

### Exercise 5 — Walk buckets from high frequency down


```python
freq = [[], [3], [2], [1], [], [], []]

for i in range(len(freq) - 1, 0, -1):
    for n in freq[i]:
        print(n)
```

### Exercise 6 — Collect until you have `k`


```python
freq = [[], [3], [2], [1], [], [], []]
k = 2
res = []

for i in range(len(freq) - 1, 0, -1):
    for n in freq[i]:
        res.append(n)
        if len(res) == k:
            break
    if len(res) == k:
        break

print(res)
```

### Exercise 9 — End-to-end on a small example


```python
nums = [1, 2, 1, 2, 1, 2, 3, 1, 3, 2]
k = 2

count = {}
for n in nums:
    count[n] = 1 + count.get(n, 0)

freq = [[] for _ in range(len(nums) + 1)]
for n, c in count.items():
    freq[c].append(n)

res = []
for i in range(len(freq) - 1, 0, -1):
    for n in freq[i]:
        res.append(n)
        if len(res) == k:
            break
    if len(res) == k:
        break

print(res)
```

### Exercise 10 — Put it in a function


```python
def top_k_frequent(nums, k):
    count = {}
    for n in nums:
        count[n] = 1 + count.get(n, 0)

    freq = [[] for _ in range(len(nums) + 1)]
    for n, c in count.items():
        freq[c].append(n)

    res = []
    for i in range(len(freq) - 1, 0, -1):
        for n in freq[i]:
            res.append(n)
            if len(res) == k:
                return res
    return res


print(top_k_frequent([1, 1, 1, 2, 2, 3], 2))
print(top_k_frequent([1], 1))
```

## Part 2 — Full solution

O(n) bucket solution — count frequencies, place numbers into buckets by count, walk from highest frequency down until you have `k` elements.


```python
class Solution:
    def topKFrequent(self, nums: list[int], k: int) -> list[int]:
        count = {}
        freq = [[] for _ in range(len(nums) + 1)]

        for n in nums:
            count[n] = 1 + count.get(n, 0)

        for n, c in count.items():
            freq[c].append(n)

        res = []
        for i in range(len(freq) - 1, 0, -1):
            for n in freq[i]:
                res.append(n)
                if len(res) == k:
                    return res
        return res
```

```python
sol = Solution()

assert sorted(sol.topKFrequent([1, 1, 1, 2, 2, 3], 2)) == [1, 2]
assert sorted(sol.topKFrequent([1], 1)) == [1]
assert sorted(sol.topKFrequent([1, 2, 1, 2, 1, 2, 3, 1, 3, 2], 2)) == [1, 2]
assert sorted(sol.topKFrequent([4, 1, -1, 2, -1, 2, 3], 2)) == [-1, 2]

print("All tests passed.")
```
