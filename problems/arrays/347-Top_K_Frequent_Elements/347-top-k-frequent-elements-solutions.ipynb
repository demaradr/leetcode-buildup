## Many ways to solve

| Approach | Idea | Time |
|----------|------|------|
| **Count + sort** | Sort unique numbers by frequency | O(n log n) |
| **Count + heap** (below) | Min-heap of size `k` | O(n log k) |
| **Count + buckets** | Index = frequency; walk high → low | O(n) |

**The heap approach is the one to practice** — O(n log k) time, O(n) space for the count dict. It satisfies the follow-up (better than O(n log n)) and transfers to many other "top k" problems.

## Why a min-heap of size `k`?

Store `(frequency, number)` so the heap orders by count. When the heap grows past `k`, pop the least frequent. Whatever remains are the top `k`.

## Part 1 — Skill exercise solutions


### Exercise 1 — Count frequencies with `.get()`


```python
nums = [1, 1, 1, 2, 2, 3]
count = {}

for n in nums:
    count[n] = 1 + count.get(n, 0)

print(count)
```

### Exercise 2 — Loop over `count.items()`


```python
count = {1: 3, 2: 2, 3: 1}

for n, c in count.items():
    print(n, c)
```

### Exercise 3 — Push `(frequency, number)` onto a heap


```python
import heapq

count = {1: 3, 2: 2, 3: 1}
heap = []

for n, c in count.items():
    heapq.heappush(heap, (c, n))

print(heap)
```

### Exercise 4 — Peek and pop the smallest


```python
import heapq

heap = [(1, 3), (3, 1), (2, 2)]

print(heap[0])
popped = heapq.heappop(heap)
print(popped)
print(heap)
```

### Exercise 5 — Keep the heap at size `k`


```python
import heapq

count = {1: 3, 2: 2, 3: 1}
k = 2
heap = []

for n, c in count.items():
    heapq.heappush(heap, (c, n))
    if len(heap) > k:
        heapq.heappop(heap)

print(heap)
```

### Exercise 6 — Extract the numbers from the heap


```python
heap = [(2, 2), (3, 1)]

res = [n for c, n in heap]
print(res)
```

### Exercise 9 — End-to-end on a small example


```python
import heapq

nums = [1, 2, 1, 2, 1, 2, 3, 1, 3, 2]
k = 2

count = {}
for n in nums:
    count[n] = 1 + count.get(n, 0)

heap = []
for n, c in count.items():
    heapq.heappush(heap, (c, n))
    if len(heap) > k:
        heapq.heappop(heap)

res = [n for c, n in heap]
print(res)
```

### Exercise 10 — Put it in a function


```python
import heapq

def top_k_frequent(nums, k):
    count = {}
    for n in nums:
        count[n] = 1 + count.get(n, 0)

    heap = []
    for n, c in count.items():
        heapq.heappush(heap, (c, n))
        if len(heap) > k:
            heapq.heappop(heap)

    return [n for c, n in heap]

print(top_k_frequent([1, 1, 1, 2, 2, 3], 2))
print(top_k_frequent([1], 1))
```

## Part 2 — Full solution

O(n log k) heap solution — count frequencies, keep a min-heap of size `k` with `(frequency, number)` pairs, return the numbers left in the heap.


```python
import heapq

class Solution:
    def topKFrequent(self, nums: list[int], k: int) -> list[int]:
        count = {}
        for n in nums:
            count[n] = 1 + count.get(n, 0)

        heap = []
        for n, c in count.items():
            heapq.heappush(heap, (c, n))
            if len(heap) > k:
                heapq.heappop(heap)

        return [n for c, n in heap]
```

```python
sol = Solution()

assert sorted(sol.topKFrequent([1, 1, 1, 2, 2, 3], 2)) == [1, 2]
assert sorted(sol.topKFrequent([1], 1)) == [1]
assert sorted(sol.topKFrequent([1, 2, 1, 2, 1, 2, 3, 1, 3, 2], 2)) == [1, 2]
assert sorted(sol.topKFrequent([4, 1, -1, 2, -1, 2, 3], 2)) == [-1, 2]

print("All tests passed.")
```
