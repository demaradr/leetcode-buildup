## Many ways to solve

| Approach | Idea | Time |
|----------|------|------|
| **Repeated replace** | Keep replacing `()`, `[]`, `{}` until nothing left | O(n²) |
| **Stack** (what you'll build) | Push opens; on a close, pop and check it matches | O(n) |

**Stack is the classic pattern** — one pass, O(n) time, O(n) space in the worst case.

## Why a stack?

The **most recent** open bracket must be closed first. A stack remembers opens in that order (LIFO):

```python
stack = []
close_to_open = {')': '(', ']': '[', '}': '{'}
```

For each character `c`:
1. If `c` is a **closing** bracket (`c in close_to_open`):
   - Stack must be non-empty and `stack[-1]` must equal `close_to_open[c]`
   - If so → `stack.pop()`; else → invalid
2. If `c` is an **opening** bracket → `stack.append(c)`
3. At the end, the stack must be **empty** (every open was closed)
