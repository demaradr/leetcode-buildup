## Many ways to solve

| Approach | Idea | Time |
|----------|------|------|
| **Clean + reverse** | Filter to alphanumeric, lowercase, compare to reverse | O(n) |
| **Two pointers** (below) | Skip junk from both ends, compare inward | O(n) |

**Two pointers is the classic pattern** — O(n) time, O(1) extra space.

## Why `is_alphanumeric`?

Instead of `.isalnum()`, we check character codes with `ord()`. A character is alphanumeric if its code falls in the uppercase, lowercase, or digit ranges.

### Exercise 8 — Move pointers inward


```python
def is_alphanumeric(c):
    return (ord('A') <= ord(c) <= ord('Z') or
            ord('a') <= ord(c) <= ord('z') or
            ord('0') <= ord(c) <= ord('9'))

s = "A man, a plan, a canal: Panama"
l, r = 0, len(s) - 1
both_match = True

for _ in range(2):
    while l < r and not is_alphanumeric(s[l]):
        l += 1
    while l < r and not is_alphanumeric(s[r]):
        r -= 1
    if s[l].lower() != s[r].lower():
        both_match = False
    l += 1
    r -= 1

print(both_match)
```

### Exercise 9 — Detect a mismatch


```python
def is_alphanumeric(c):
    return (ord('A') <= ord(c) <= ord('Z') or
            ord('a') <= ord(c) <= ord('z') or
            ord('0') <= ord(c) <= ord('9'))

s = "race a car"
l, r = 0, len(s) - 1

while l < r and not is_alphanumeric(s[l]):
    l += 1
while l < r and not is_alphanumeric(s[r]):
    r -= 1

if s[l].lower() != s[r].lower():
    print("mismatch")
```

### Exercise 10 — Put it in a function


```python
def is_alphanumeric(c):
    return (ord('A') <= ord(c) <= ord('Z') or
            ord('a') <= ord(c) <= ord('z') or
            ord('0') <= ord(c) <= ord('9'))

def is_palindrome(s):
    l, r = 0, len(s) - 1

    while l < r:
        while l < r and not is_alphanumeric(s[l]):
            l += 1
        while l < r and not is_alphanumeric(s[r]):
            r -= 1

        if s[l].lower() != s[r].lower():
            return False

        l += 1
        r -= 1

    return True


print(is_palindrome("A man, a plan, a canal: Panama"))
print(is_palindrome("race a car"))
print(is_palindrome(" "))
```

## Part 2 — Full solution

O(n) two-pointer solution with a custom `is_alphanumeric` helper.


```python
def is_alphanumeric(c):
    return (ord('A') <= ord(c) <= ord('Z') or
            ord('a') <= ord(c) <= ord('z') or
            ord('0') <= ord(c) <= ord('9'))

class Solution:
    def isPalindrome(self, s: str) -> bool:
        l, r = 0, len(s) - 1

        while l < r:
            while l < r and not is_alphanumeric(s[l]):
                l += 1
            while l < r and not is_alphanumeric(s[r]):
                r -= 1

            if s[l].lower() != s[r].lower():
                return False

            l += 1
            r -= 1

        return True
```

```python
sol = Solution()

assert sol.isPalindrome("A man, a plan, a canal: Panama") is True
assert sol.isPalindrome("race a car") is False
assert sol.isPalindrome(" ") is True
assert sol.isPalindrome("0P") is False
assert sol.isPalindrome("a.") is True

print("All tests passed.")
```
