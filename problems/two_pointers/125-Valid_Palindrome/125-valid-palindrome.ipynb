## Many ways to solve

| Approach | Idea | Time |
|----------|------|------|
| **Clean + reverse** | Strip non-alphanumeric, lowercase, compare to its reverse | O(n) |
| **Two pointers** (what you'll build) | `l` at start, `r` at end; skip junk, compare letters, move inward | O(n) |

Both are O(n). **Two pointers is the classic pattern** — you compare in place without building a new string.

## Why two pointers?

A palindrome reads the same forward and backward. Put one pointer at the **left** and one at the **right**:

```python
l, r = 0, len(s) - 1
```

On each step:
1. Skip characters that are **not** letters or digits (spaces, punctuation, etc.)
2. Compare `s[l]` and `s[r]` case-insensitively (use `.lower()`)
3. If they differ → not a palindrome
4. Move `l` right and `r` left
5. Stop when `l >= r`

You'll write your own helper instead of using `.isalnum()`:

```python
def is_alphanumeric(c):
    return (ord('A') <= ord(c) <= ord('Z') or
            ord('a') <= ord(c) <= ord('z') or
            ord('0') <= ord(c) <= ord('9'))
```
